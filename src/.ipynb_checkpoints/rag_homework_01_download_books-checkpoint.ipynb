## Step 1 — Imports and settings

```python
from pathlib import Path
from urllib.parse import urlparse

import pandas as pd
import requests
from IPython.display import display

CSV_URL = (
    "https://raw.githubusercontent.com/alexeygrigorev/"
    "ai-engineering-buildcamp-code/main/01-foundation/homework/books.csv"
)
CSV_PATH = Path("books.csv")
BOOKS_DIR = Path("books")

# Some servers reject requests that have no User-Agent
HEADERS = {"User-Agent": "Mozilla/5.0 (homework downloader)"}
TIMEOUT = 60  # seconds

BOOKS_DIR.mkdir(exist_ok=True)

print("Working directory:", Path.cwd())
print("CSV will be saved to:", CSV_PATH.resolve())
print("PDFs will be saved to:", BOOKS_DIR.resolve())
```

## Step 2 — Download `books.csv`

Same result as the `wget` command in the homework, but it works on Windows too.

```python
response = requests.get(CSV_URL, headers=HEADERS, timeout=TIMEOUT)
response.raise_for_status()
CSV_PATH.write_bytes(response.content)

print(f"HTTP status : {response.status_code}")
print(f"Saved       : {CSV_PATH}  ({CSV_PATH.stat().st_size:,} bytes)")
print()
print("Raw file content:")
print(CSV_PATH.read_text(encoding="utf-8"))
```

## Step 3 — Load and inspect the CSV

**Check:** 7 rows, 3 columns (`title`, `book_url`, `pdf_url`), no missing values, and the first row is *Think Python 2e* pointing at `thinkpython2.pdf` (the 2nd edition the homework questions refer to).

```python
books = pd.read_csv(CSV_PATH)

# The file name each PDF will get on disk = last part of its URL
books["filename"] = books["pdf_url"].apply(lambda url: Path(urlparse(url).path).name)

print("Shape (rows, columns):", books.shape)
print("Columns:", list(books.columns))
print("Missing values:", int(books.isna().sum().sum()))
print("Duplicate file names:", int(books["filename"].duplicated().sum()))

with pd.option_context("display.max_colwidth", None):
    display(books[["title", "filename", "pdf_url"]])
```

## Step 4 — Define the download function

It streams each file to disk in pieces, writes to a temporary `.part` file first (so an interrupted download never leaves a broken PDF behind), and skips books that are already downloaded, so the cell is safe to re-run.

```python
def is_pdf(path):
    """A real PDF starts with the bytes %PDF."""
    with open(path, "rb") as f:
        return f.read(4) == b"%PDF"


def download_pdf(url, target, overwrite=False):
    """Download one PDF. Returns a status string: downloaded / skipped / error."""
    if target.exists() and not overwrite and is_pdf(target):
        return "skipped (already there)"

    part = target.with_name(target.name + ".part")
    try:
        with requests.get(url, headers=HEADERS, timeout=TIMEOUT, stream=True) as r:
            r.raise_for_status()
            with open(part, "wb") as f:
                for chunk in r.iter_content(chunk_size=1024 * 256):
                    f.write(chunk)

        if not is_pdf(part):
            part.unlink()
            return "error: server did not return a PDF"

        part.replace(target)
        return "downloaded"
    except requests.RequestException as e:
        part.unlink(missing_ok=True)
        code = getattr(e.response, "status_code", None)
        detail = f"HTTP {code}" if code else str(e)[:80]
        return f"error: {type(e).__name__} ({detail})"


print("Functions defined: is_pdf, download_pdf")
```

## Step 5 — Download all the books

This can take a minute or two. Each book prints one line as it finishes.

```python
statuses = []

for i, row in enumerate(books.itertuples(index=False), start=1):
    target = BOOKS_DIR / row.filename
    print(f"[{i}/{len(books)}] {row.title:<30} ", end="", flush=True)

    status = download_pdf(row.pdf_url, target)
    statuses.append(status)

    size = f"{target.stat().st_size / 1_048_576:6.2f} MB" if target.exists() else "   --    "
    print(f"{size}  {status}")

books["status"] = statuses
print()
print("Done.")
```

## Step 6 — Verify the downloads

**Check:** every row has `valid_pdf = True` and a sensible size (roughly 0.5–5 MB each). The last line should say all 7 books are ready.

```python
def describe(filename):
    path = BOOKS_DIR / filename
    if not path.exists():
        return pd.Series({"size_mb": None, "valid_pdf": False})
    return pd.Series({
        "size_mb": round(path.stat().st_size / 1_048_576, 2),
        "valid_pdf": is_pdf(path),
    })


summary = pd.concat([books[["title", "filename", "status"]], books["filename"].apply(describe)], axis=1)

with pd.option_context("display.max_colwidth", None):
    display(summary)

ok = int(summary["valid_pdf"].sum())
print(f"Valid PDFs : {ok} of {len(summary)}")
print(f"Total size : {summary['size_mb'].sum():.2f} MB")

if ok == len(summary):
    print("All books are ready - you can move on to text extraction.")
else:
    failed = summary.loc[~summary["valid_pdf"], "title"].tolist()
    print("Still missing:", failed)
    print("Re-run Step 5 - it only retries the books that are missing.")
```

## Step 7 — Final check of the folder

Lists what is really on disk and confirms the Think Python file is the 2nd edition.

```python
files = sorted(BOOKS_DIR.glob("*.pdf"))
print(f"{len(files)} PDF files in {BOOKS_DIR.resolve()}:")
for path in files:
    print(f"  {path.name:<34} {path.stat().st_size:>12,} bytes")

leftovers = list(BOOKS_DIR.glob("*.part"))
print()
print("Unfinished .part files:", [p.name for p in leftovers] or "none")

think_python = BOOKS_DIR / "thinkpython2.pdf"
print("Think Python 2nd edition present:", think_python.exists())
```

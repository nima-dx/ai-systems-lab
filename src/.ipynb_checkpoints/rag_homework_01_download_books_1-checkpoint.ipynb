## Step 1 — Imports and settings

````python
from pathlib import Path
from urllib.parse import urlparse

import pandas as pd
import requests
from IPython.display import display

CSV_URL = (
    "https://raw.githubusercontent.com/alexeygrigorev/"
    "ai-engineering-buildcamp-code/main/01-foundation/homework/books.csv"
)
CSV_PATH = Path("books.csv")
BOOKS_DIR = Path("books")

# Some servers reject requests that have no User-Agent
HEADERS = {"User-Agent": "Mozilla/5.0 (homework downloader)"}
TIMEOUT = 60  # seconds

BOOKS_DIR.mkdir(exist_ok=True)

print("Working directory:", Path.cwd())
print("CSV will be saved to:", CSV_PATH.resolve())
print("PDFs will be saved to:", BOOKS_DIR.resolve())
````

## Step 2 — Download `books.csv`

Same result as the `wget` command in the homework, but it works on Windows too.

````python
response = requests.get(CSV_URL, headers=HEADERS, timeout=TIMEOUT)
response.raise_for_status()
CSV_PATH.write_bytes(response.content)

print(f"HTTP status : {response.status_code}")
print(f"Saved       : {CSV_PATH}  ({CSV_PATH.stat().st_size:,} bytes)")
print()
print("Raw file content:")
print(CSV_PATH.read_text(encoding="utf-8"))
````

## Step 3 — Load and inspect the CSV

**Check:** 7 rows, 3 columns (`title`, `book_url`, `pdf_url`), no missing values, and the first row is *Think Python 2e* pointing at `thinkpython2.pdf` (the 2nd edition the homework questions refer to).

````python
books = pd.read_csv(CSV_PATH)

# The file name each PDF will get on disk = last part of its URL
books["filename"] = books["pdf_url"].apply(lambda url: Path(urlparse(url).path).name)

print("Shape (rows, columns):", books.shape)
print("Columns:", list(books.columns))
print("Missing values:", int(books.isna().sum().sum()))
print("Duplicate file names:", int(books["filename"].duplicated().sum()))

with pd.option_context("display.max_colwidth", None):
    display(books[["title", "filename", "pdf_url"]])
````

## Step 4 — Define the download function

It streams each file to disk in pieces, writes to a temporary `.part` file first (so an interrupted download never leaves a broken PDF behind), and skips books that are already downloaded, so the cell is safe to re-run.

````python
def is_pdf(path):
    """A real PDF starts with the bytes %PDF."""
    with open(path, "rb") as f:
        return f.read(4) == b"%PDF"


def download_pdf(url, target, overwrite=False):
    """Download one PDF. Returns a status string: downloaded / skipped / error."""
    if target.exists() and not overwrite and is_pdf(target):
        return "skipped (already there)"

    part = target.with_name(target.name + ".part")
    try:
        with requests.get(url, headers=HEADERS, timeout=TIMEOUT, stream=True) as r:
            r.raise_for_status()
            with open(part, "wb") as f:
                for chunk in r.iter_content(chunk_size=1024 * 256):
                    f.write(chunk)

        if not is_pdf(part):
            part.unlink()
            return "error: server did not return a PDF"

        part.replace(target)
        return "downloaded"
    except requests.RequestException as e:
        part.unlink(missing_ok=True)
        code = getattr(e.response, "status_code", None)
        detail = f"HTTP {code}" if code else str(e)[:80]
        return f"error: {type(e).__name__} ({detail})"


print("Functions defined: is_pdf, download_pdf")
````

## Step 5 — Download all the books

This can take a minute or two. Each book prints one line as it finishes.

````python
statuses = []

for i, row in enumerate(books.itertuples(index=False), start=1):
    target = BOOKS_DIR / row.filename
    print(f"[{i}/{len(books)}] {row.title:<30} ", end="", flush=True)

    status = download_pdf(row.pdf_url, target)
    statuses.append(status)

    size = f"{target.stat().st_size / 1_048_576:6.2f} MB" if target.exists() else "   --    "
    print(f"{size}  {status}")

books["status"] = statuses
print()
print("Done.")
````

## Step 6 — Verify the downloads

**Check:** every row has `valid_pdf = True` and a sensible size (roughly 0.5–5 MB each). The last line should say all 7 books are ready.

````python
def describe(filename):
    path = BOOKS_DIR / filename
    if not path.exists():
        return pd.Series({"size_mb": None, "valid_pdf": False})
    return pd.Series({
        "size_mb": round(path.stat().st_size / 1_048_576, 2),
        "valid_pdf": is_pdf(path),
    })


summary = pd.concat([books[["title", "filename", "status"]], books["filename"].apply(describe)], axis=1)

with pd.option_context("display.max_colwidth", None):
    display(summary)

ok = int(summary["valid_pdf"].sum())
print(f"Valid PDFs : {ok} of {len(summary)}")
print(f"Total size : {summary['size_mb'].sum():.2f} MB")

if ok == len(summary):
    print("All books are ready - you can move on to text extraction.")
else:
    failed = summary.loc[~summary["valid_pdf"], "title"].tolist()
    print("Still missing:", failed)
    print("Re-run Step 5 - it only retries the books that are missing.")
````

## Step 7 — Final check of the folder

Lists what is really on disk and confirms the Think Python file is the 2nd edition.

````python
files = sorted(BOOKS_DIR.glob("*.pdf"))
print(f"{len(files)} PDF files in {BOOKS_DIR.resolve()}:")
for path in files:
    print(f"  {path.name:<34} {path.stat().st_size:>12,} bytes")

leftovers = list(BOOKS_DIR.glob("*.part"))
print()
print("Unfinished .part files:", [p.name for p in leftovers] or "none")

think_python = BOOKS_DIR / "thinkpython2.pdf"
print("Think Python 2nd edition present:", think_python.exists())
````

## Step 9 — Extract text from every PDF with `markitdown`

`markitdown` converts each PDF to markdown-ish plain text. This is slower than the download step — expect roughly 5-15 seconds per book.

Each resulting `documents` entry is a dict with `title`, `filename`, and `content` (matching the `title` / `description` / `content` shape used in the course's RAG lesson, minus `description`, which PDFs don't have).

````python
import time

from markitdown import MarkItDown

converter = MarkItDown()
documents = []

pdf_files = sorted(BOOKS_DIR.glob("*.pdf"))
title_by_filename = dict(zip(books["filename"], books["title"]))

for i, path in enumerate(pdf_files, start=1):
    print(f"[{i}/{len(pdf_files)}] {path.name:<34} ", end="", flush=True)
    start = time.time()

    result = converter.convert(str(path))
    content = result.text_content

    documents.append({
        "title": title_by_filename.get(path.name, path.stem),
        "filename": path.name,
        "content": content,
    })

    elapsed = time.time() - start
    print(f"{elapsed:5.1f}s  {len(content):>8,} chars")

print()
print(f"Extracted {len(documents)} documents.")
````

## Step 10 — Inspect the extracted content

**This is where you answer questions like:** *"How many lines are in the extracted content from the Think Python book (2nd edition)?"*

A "line" is whatever comes out of splitting the text on newlines — `content.splitlines()` (equivalently `content.count(chr(10)) + 1`, as long as the text doesn't end with a trailing newline that would make `splitlines()` 1 shorter; both are shown below so you can compare).

````python
rows = []
for doc in documents:
    content = doc["content"]
    rows.append({
        "title": doc["title"],
        "filename": doc["filename"],
        "characters": len(content),
        "lines_splitlines": len(content.splitlines()),
        "lines_newline_count_plus_1": content.count("\n") + 1,
    })

extraction_summary = pd.DataFrame(rows)
with pd.option_context("display.max_colwidth", None):
    display(extraction_summary)
````

````python
# Direct answer for Think Python 2e
think_python_doc = next(d for d in documents if "think python" in d["title"].lower())
tp_content = think_python_doc["content"]

print("Title   :", think_python_doc["title"])
print("Filename:", think_python_doc["filename"])
print("Characters:", len(tp_content))
print("Lines (splitlines):", len(tp_content.splitlines()))
print("Lines (newline count + 1):", tp_content.count("\n") + 1)
print()
print("First 5 lines:")
for line in tp_content.splitlines()[:5]:
    print(" ", repr(line))
````

## Step 11 — Index the full documents with `minsearch`

Same pattern as the course's RAG lesson: build an `Index` over the text fields, `fit` it on the documents, and try a search.

````python
from minsearch import Index

doc_index = Index(
    text_fields=["title", "content"],
    keyword_fields=["filename"],
)
doc_index.fit(documents)

print("Indexed", len(documents), "documents.")

sample_query = "what is a variable"
results = doc_index.search(sample_query, num_results=3)

print(f"Top {len(results)} results for: {sample_query!r}")
for r in results:
    print(f"  - {r['title']}  ({len(r['content']):,} chars)")
````

## Step 12 — Chunk the documents

Full books are far too long to put in a single prompt, so they get split into overlapping chunks with `gitsource`'s `chunk_documents` (a thin wrapper over the sliding-window chunker from the course lesson: `size` is the chunk length in characters, `step` is how far the window moves each time, so `size - step` is the overlap between consecutive chunks).

````python
from gitsource import chunk_documents

CHUNK_SIZE = 2000
CHUNK_STEP = 1000

document_chunks = chunk_documents(documents, size=CHUNK_SIZE, step=CHUNK_STEP)

print(f"{len(documents)} documents -> {len(document_chunks)} chunks")
print(f"(size={CHUNK_SIZE}, step={CHUNK_STEP}, overlap={CHUNK_SIZE - CHUNK_STEP} chars)")
print()
print("Example chunk:")
display(document_chunks[5])
````

````python
chunk_counts = pd.Series([c["title"] for c in document_chunks]).value_counts()
print("Chunks per book:")
display(chunk_counts)
````

## Step 13 — Index the chunks

Rebuild the `minsearch` index, this time over the chunks instead of the whole documents, so search results are short enough to fit in a prompt.

````python
chunk_index = Index(
    text_fields=["title", "content"],
    keyword_fields=["filename"],
)
chunk_index.fit(document_chunks)

print("Indexed", len(document_chunks), "chunks.")

results = chunk_index.search(sample_query, num_results=3)
print(f"Top {len(results)} chunk results for: {sample_query!r}")
for r in results:
    print(f"  - {r['title']}  start={r['start']:>6}  ({len(r['content'])} chars)")
    print("    ", repr(r["content"][:120]))
````

## Step 14 — `search`, `build_prompt`, `llm`, and `rag(query)`

Same three-step pattern as the course's RAG lesson, using the OpenAI `responses` API (`openai_client.responses.create`, reading `response.output_text`) instead of the older chat-completions API.

````python
import json

from openai import OpenAI

openai_client = OpenAI()

RAG_INSTRUCTIONS = """
You're a helpful assistant answering questions about a set of programming books
by Allen Downey (Green Tea Press). Answer the QUESTION using only the CONTEXT
below. If the answer isn't in the CONTEXT, say you don't know.
""".strip()


def search(query, num_results=5):
    return chunk_index.search(query, num_results=num_results)


def build_prompt(query, search_results):
    context = json.dumps(search_results, indent=2)
    return f"""
<QUESTION>
{query}
</QUESTION>

<CONTEXT>
{context}
</CONTEXT>
""".strip()


def llm(prompt, instructions=RAG_INSTRUCTIONS, model="gpt-4o-mini"):
    response = openai_client.responses.create(
        model=model,
        input=[
            {"role": "system", "content": instructions},
            {"role": "user", "content": prompt},
        ],
    )
    return response.output_text


def rag(query):
    search_results = search(query)
    prompt = build_prompt(query, search_results)
    answer = llm(prompt)
    return answer


print("search, build_prompt, llm, rag all defined.")
````

## Step 15 — Try it out

Run a query through the full pipeline and look at each intermediate step.

````python
test_query = "What is a variable in Python?"

step1_results = search(test_query)
print(f"search() returned {len(step1_results)} chunks:")
for r in step1_results:
    print(f"  - {r['title']}  ({len(r['content'])} chars)")

print()
step2_prompt = build_prompt(test_query, step1_results)
print("build_prompt() produced a prompt of", len(step2_prompt), "characters. First 300 chars:")
print(step2_prompt[:300])
````

````python
step3_answer = rag(test_query)
print("Question:", test_query)
print()
print("Answer:")
print(step3_answer)
````
